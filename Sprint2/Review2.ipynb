```
public class Employee
{
    private string _name;
    private string _position;
    private float _salary;
    
    public Employee(string name, string position, float salary)
    {
        _name = name;
        _position = position;
        _salary = salary;
        
    }

    public string Name
    {
        get { return _name; }
        set 
        {
            if (!string.IsNullOrWhiteSpace(value))
                _name = value;
            else
                throw new ArgumentException("Имя не может быть пустым");
        }
    }

    public string Position
    {
        get { return _position; }
        set 
        {
            if (!string.IsNullOrWhiteSpace(value))
                _position = value;
            else
                throw new ArgumentException("Должность не может быть пустой");
        }
    }

    public float Salary
    {
        get { return _salary; }
        set 
        {
            if (value >= 0)
                _salary = value;
            else
                throw new ArgumentException("Зарплата не может быть отрицательной");
        }
    }

    

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Имя: {Name}. Возраст: {Position}. Зарплата: {Salary}.");
    }
}

public class Manager : Employee
{
    private string _department;
    

    public Manager(string name, string position, float salary, string department) 
        : base(name, position, salary)
    {
        _department = department;
    }

    public string Department
    {
        get { return _department; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _department = value;
            else
                throw new ArgumentException("Название отдела не может быть пустым");
        }
    }

    

    public override void PrintInfo()
    {
        Console.WriteLine($"Имя: {Name}. Должность: {Position}. Зарплата: {Salary}. Отдел: {Department}.");
    }

    
}


public interface IReportable
{
    void GenerateReport();
}

public interface ITeamLeader
{
    void GetTeamMembers(string[] Members);
}



public class ProjectManager : Manager, IReportable, ITeamLeader
{
    private string _projectName;

    public ProjectManager(string name, string position, float salary, string department, string projectName) 
        : base(name, position, salary, department)
    {
        _projectName = projectName;
    }

    public string ProjectName
    {
        get { return _projectName; }
        set
        {
            if (!string.IsNullOrWhiteSpace(value))
                _projectName = value;
            else
                throw new ArgumentException("Название проекта не может быть пустым");
        }
    }


    public override void PrintInfo()
    {
        Console.WriteLine($"Имя: {Name}. Должность: {Position}. Зарплата: {Salary}. Отдел: {Department}. Проект: {ProjectName}.");
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Отчет по проекту: {ProjectName}");
    }

    public void GetTeamMembers(string[] Members)
    {
        Console.WriteLine($"Члены команды проекта {ProjectName}:");
        for (int i = 0; i < Members.Length; i++)
        {
            Console.WriteLine(Members[i]);
        }
    }

}



Console.WriteLine("=== ПРОСТОЕ НАСЛЕДОВАНИЕ ===");
        
        Manager manager = new Manager("Ник", "Аналитик", 80000, "IT");
        
        Console.WriteLine("Информация о менеджере:");
        manager.PrintInfo();

        Console.WriteLine("\n=== СЛОЖНОЕ НАСЛЕДОВАНИЕ ===");
        Console.WriteLine("=== МНОЖЕСТВЕННОЕ НАСЛЕДОВАНИЕ ===");

        ProjectManager projectManager = new ProjectManager("Мария", "Дизайнер", 95000, "Web", "Введение в микросервисы");
        
        Console.WriteLine("Информация о проектном менеджере:");
        projectManager.PrintInfo();
        
        Console.WriteLine("\nГенерация отчета:");
        projectManager.GenerateReport();
        
        Console.WriteLine("\nЧлены команды:");
        string[] teamMembers = {"Анна", "Петр", "Сергей"};
        projectManager.GetTeamMembers(teamMembers);
```
