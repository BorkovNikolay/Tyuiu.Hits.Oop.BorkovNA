```
using System;
using System.Collections.Generic;

public interface IBankAccount
{
    void GetInfo();
    decimal Deposit(decimal amount);
    decimal Withdraw(decimal amount);
}

public delegate void DisplayDelegate();

public class BankAccount: IBankAccount
{

    private string accountNumber;
    private decimal balance;
    private string accountType;

    public delegate void AccountHandler(string message);
    public event AccountHandler AccountUpdated;

    public string AccountNumber 
    {
        get {return accountNumber;}
        set {accountNumber = value;}
    }

    public decimal Balance 
    {
        get {return balance;}
        set {balance = value; }
    }

    public string AccountType 
    {
        get {return accountType;}
        set {accountType = value;}
    }

    public void GetInfo() 
    {
        Console.WriteLine($"Номер счета: {AccountNumber}. Тип счета: {AccountType}. Баланс: {Balance}.");
    }

    public void UpdateModel(string newAccountType)
    {
        AccountType = newAccountType;
        AccountUpdated?.Invoke($"Тип счета изменен на {newAccountType}"); 
    }

    public virtual decimal Deposit(decimal amount) 
    {
        balance = balance + amount;
        return balance;
    }

    public virtual decimal Withdraw(decimal amount) 
    {
        balance = balance - amount;
        return balance;
    }
}

public class SavingsAccount: BankAccount, IBankAccount
{
    private decimal interestRate;

    public SavingsAccount(): base()
    {
        AccountNumber = "00000000";
        AccountType = "Сберегательный счет";
        Balance = 200m;
        interestRate = 0.4m;
    }

    public decimal InterestRate 
    {
        get{return interestRate;}
        set{interestRate = value;}
    }

    public override decimal Deposit(decimal amount) 
    {
        Balance = Balance + amount * interestRate;
        return Balance;
    }

    public override decimal Withdraw(decimal amount) 
    {
        Balance = Balance - amount;
        return Balance;
    }
}

public class CheckingAccount: BankAccount, IBankAccount
{
    private decimal overdraftLimit;

    public CheckingAccount(): base()
    {
        AccountNumber = "00000001";
        AccountType = "Текущий счет";
        Balance = 200m;
        OverdraftLimit = 100m;
    }

    public decimal OverdraftLimit 
    {
        get{return overdraftLimit;}
        set{overdraftLimit = value;}
    }

    public override decimal Withdraw(decimal amount) 
    {
        if (OverdraftLimit > amount) {
            Balance = Balance - amount;
            return Balance;
        }
        else {
            Console.WriteLine("Слишком большая сумма для снятия наличных");
            return Balance;
        }
    }
}

public class AccountCollection<T> where T : BankAccount
{
    private List<T> _accounts = new List<T>();

    public void Add(T account)
    {
        _accounts.Add(account);
    }

    public void Remove(T account)
    {
        _accounts.Remove(account);
    }

    public void DisplayAccountInfo()
    {
        foreach (var account in _accounts)
        {
            account.GetInfo();
        }
    }

    public void DepositAccounts()
    {
        foreach (var account in _accounts)
        {
            account.Deposit(100m);
        }
    }

    public void WithdrawAccounts()
    {
        foreach (var account in _accounts)
        {
            account.Withdraw(50m);
        }
    }
}

public class BankWindow
{
    private readonly IBankAccount _bankAccount;

    public BankWindow(IBankAccount bankAccount)
    {
        _bankAccount = bankAccount; 
    }

    public void WithdrawAndDeposit()
    {
        _bankAccount.Withdraw(10m);
        _bankAccount.Deposit(20m);
        _bankAccount.GetInfo();
    }
}
Console.WriteLine("");
AccountCollection<BankAccount> accounts = new AccountCollection<BankAccount>();
accounts.Add(new SavingsAccount());
accounts.Add(new CheckingAccount());
accounts.DisplayAccountInfo();
accounts.DepositAccounts();
accounts.WithdrawAccounts();
accounts.DisplayAccountInfo();
Console.WriteLine("");

IBankAccount mySavingsAccount = new SavingsAccount(); 
BankWindow SavingsAccountWindow = new BankWindow(mySavingsAccount);
SavingsAccountWindow.WithdrawAndDeposit();

IBankAccount myCheckingAccount = new CheckingAccount();
BankWindow CheckingAccountWindow = new BankWindow(myCheckingAccount);
CheckingAccountWindow.WithdrawAndDeposit();
Console.WriteLine("");

SavingsAccount mySavingsAccount2 = new SavingsAccount();
CheckingAccount myCheckingAccount2 = new CheckingAccount();

DisplayDelegate displaySavingsAccount = mySavingsAccount2.GetInfo;
DisplayDelegate displayCheckingAccount = myCheckingAccount2.GetInfo;

displaySavingsAccount();
displayCheckingAccount();
Console.WriteLine("");

BankAccount bankAccount = new BankAccount ();
bankAccount.AccountUpdated += OnTransportUpdated;
bankAccount.UpdateModel("Сберегательный счет");


static void OnTransportUpdated(string message)
{
    Console.WriteLine(message);
}
```
